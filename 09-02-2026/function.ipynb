```python
def fibonnaci(n, result):
    if n == 0:
        return 0
    if n == 1:
        return 1
    
    value = (fibonnaci(n -1, result) + fibonnaci(n - 2, result))
    result.append(value)
    return value
    
    
result = []
fibonnaci(15, result)
print(result)
```
