```python
import re
```

```python
a = "How are you  doing  today?(*(^^^&E^*(E(()))))"
# Preprocessing the string to remove special characters
preprocessed = re.split(r'([,.:;?_!"()।\']|--|\s)', a)
# print(preprocessed)
preprocessed = [item.strip() for item in preprocessed if item.strip()]
print(preprocessed)
```

```python
vocab = {token: integer for integer, token in enumerate(preprocessed)}
print(vocab)
```

```python
for key,value in vocab.items():
    print(value,",",key)
    
    new = value
    value = key
    key = new
        
    
```

```python
for item in vocab:
    new = item.key
    item.value = item.key
    item.key = new
    
```
