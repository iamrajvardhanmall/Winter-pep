```python
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
data = sns.load_dataset('titanic')
```

```python
data.head()
```

```python

data['embark_town'] = data['embark_town'].astype('category')
data.info()
```

```python
data['new_col'] = data['embark_town'].str[0]
```

```python
data.head()
```

```python
data.shape
```

```python
count = 0
for i in range(len(data)):
    if data['embarked'].iloc[i] != data['new_col'].iloc[i]:
        count = count + 1
        print((data['embark_town'].iloc[i]))
        print((data['embarked']).iloc[i])
            
print(count)
```

```python
mismatch = data[data['embarked'] != data['new_col']]
print(len(mismatch))
```

```python
list1 = data['embarked'].unique()
list2 = data['embark_town'].unique()

mapping = dict(zip(list1,list2))
mapping
print(data.shape)

matched = data[data['embarked'].map(mapping) == data['embark_town']]
print(matched.shape)


not_matched = data.loc[data['embarked'].map(mapping) != data['embark_town']]
print(not_matched)
```
